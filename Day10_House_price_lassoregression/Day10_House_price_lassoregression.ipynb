Now will split the features and target

```python
from sklearn.model_selection import train_test_split
```

```python
df.head(1)
```

```python
X = df.drop(df[['SalePrice','Log_SalePrice']], axis=1)


y = df['Log_SalePrice']

print('Shape of x: ', X.shape)
print('Shape of y: ', y.shape)
```

now we will make the numeric and categorical features


```python
numeric_features = X.select_dtypes(
    include = ['float64','int64']
).columns

categorical_features = X.select_dtypes(
    include = ['str']
).columns

print(len(numeric_features))
print(len(categorical_features))
```

now we make the numeric and categorical pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
numericpp = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categoricalpp = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown="ignore"))
])
```

now we make the preprocessor


```python
prepro = ColumnTransformer([
    ('num', numericpp,numeric_features),
    ('cat', categoricalpp,categorical_features)
])
```

```python
from sklearn.linear_model import Lasso
```

```python
lassopp = Pipeline([
    ('preprocessor', prepro),
    ('model', Lasso(alpha=0.1) )
])
```

pipelines and preprocessors are ready now we will train and split the features and target

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=.20, random_state=42
)

print('Shape of X_train', X_train.shape)
print('Shape of X_test', X_test.shape)
print('Shape of y_train', y_train.shape)
print('Shape of y_test', y_test.shape)
```

```python
lassopp.fit(X_train,y_train)
```

```python
lassp_pred = lassopp.predict(X_test)
```

Now we will use the metrics

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    root_mean_squared_error,
    r2_score,
)
```

```python
print("R2 Score: ", r2_score(y_test, lassp_pred))
print("MAE Score: ", mean_absolute_error(y_test, lassp_pred))
print("MSE Score: ", mean_squared_error(y_test, lassp_pred))
print("RMSE Score: ", root_mean_squared_error(y_test, lassp_pred))
```

```python
lasso_prepro = lassopp.named_steps['preprocessor']

feature_names = lasso_prepro.get_feature_names_out()
```

```python
lasso_model = lassopp.named_steps['model']
```

```python
coefdf = pd.DataFrame({
    'feature_names' : feature_names,
    'coefficients' : lasso_model.coef_[0]
})
```

```python
coefdf.head(10).sort_values(by='coefficients', ascending= True)
```
