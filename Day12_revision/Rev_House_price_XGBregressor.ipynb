```python
X = df.drop(df[['SalePrice','Log_SalePrice']], axis=1)

y = df['Log_SalePrice']
```

```python
df.dtypes
```

```python
numeric_features = X.select_dtypes(
    include = ['int64', 'float64']
).columns

categorical_features = X.select_dtypes(
    include = ['str']
).columns
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
```

```python
numeric_pp = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_pp = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
from sklearn.compose import ColumnTransformer
```

```python
prepro = ColumnTransformer([
    ('num', numeric_pp, numeric_features),
    ('cat', categorical_pp, categorical_features)
])
```

```python
from xgboost import XGBRegressor
```

```python
XGB_PP = Pipeline([
    ('preprocessor', prepro),
    ('model', XGBRegressor(
        n_estimators = 100,
        random_state = 42,
        learning_rate = 0.1,
        max_depth = 3
    ))
])
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,y,test_size=.20, random_state=42
)
```

```python
XGB_PP.fit(X_train, y_train)
```

```python
XGB_pred = XGB_PP.predict(X_test)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    root_mean_squared_error,
    r2_score
)

print('R2 Score: ', r2_score(y_test, XGB_pred))
print('MAE: ', mean_absolute_error(y_test, XGB_pred)),
print('MSE: ', mean_squared_error(y_test,XGB_pred))
print('RMSE: ', root_mean_squared_error(y_test,XGB_pred))
```
