```python
X = df.drop(df[['SalePrice','Log_SalePrice']], axis=1)

y = df['Log_SalePrice']
```

```python
numeric_feature = X.select_dtypes(
    include = ['int64', 'float64']
).columns

categorical_features = X.select_dtypes(
    include = ['str']
).columns
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
numeric_pp = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

catergorical_pp = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
prepro = ColumnTransformer([
    ('num', numeric_pp, numeric_feature),
    ('cat', catergorical_pp, categorical_features)
])
```

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
GBR_pp = Pipeline([
    ('preprocessor',  prepro),
    ('model', GradientBoostingRegressor(
        random_state=42,
        n_estimators=100,
        learning_rate=0.1
    ))
])
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test, y_train, y_test = train_test_split(
    X,y,test_size=.20, random_state=42
)
```

```python
GBR_pp.fit(X_train,y_train)
```

```python
GBR_pred = GBR_pp.predict(X_test)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    root_mean_squared_error,
    r2_score
)

print('R2 score: ', r2_score(y_test, GBR_pred))
print('MAE: ', mean_absolute_error(y_test, GBR_pred))
print('MSE: ', mean_squared_error(y_test,GBR_pred))
print('RMSE: ', root_mean_squared_error(y_test, GBR_pred))

```

```python
GBR_steps = GBR_pp.named_steps['model']
```

```python
importance = GBR_steps.feature_importances_
```

```python
feature_names = GBR_pp.named_steps['preprocessor'].get_feature_names_out()
```

```python
featureImpdf = pd.DataFrame({
    'Featurename' : feature_names,
    'featureimportance' : importance
})
```

```python
top10 = featureImpdf.head(10).sort_values(by='featureimportance', ascending=False)
```

```python
plt.Figure(figsize=(8,6))
plt.barh(top10['Featurename'],top10['featureimportance'])
plt.title('Top 10 features')
plt.xlabel("Importance")
plt.ylabel("Features")
plt.gca().invert_yaxis()
plt.show()
```

Now we will do permutation


```python
from sklearn.inspection import permutation_importance
```

```python
perm_importance = permutation_importance(
    GBR_pp,
    X_test,
    y_test,
    scoring='r2',
    n_repeats=10,
    random_state=42
)
```

```python
permdf = pd.DataFrame({
    'pfeaturenames': X_test.columns,
    'pimportance': perm_importance.importances_mean
})
```

```python
top10p = permdf.head(10).sort_values(by='pimportance', ascending=False)
```

```python
plt.figure(figsize=(8,6))
plt.barh(top10p['pfeaturenames'], top10p['pimportance'])
plt.title('Top 10 features after permutations')
plt.xlabel('Importance')
plt.ylabel('Features')
plt.gca().invert_yaxis()
plt.show()
```
